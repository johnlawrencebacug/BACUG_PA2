# **REPRODUCIBLE NORMALIZATION PROBLEM**

```python
import numpy as np
```

```python
np.random.seed(2112)
```

```python
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
x_bar = np.mean(X)
x_bar
```

```python
sigma = np.std(X)
sigma
```

```python
X_normalized = (X - x_bar)/sigma
X_normalized
```

```python
np.std(X_normalized)
```

```python
np.mean(X_normalized)
```

```python
np.save('X_normalized.npy', X_normalized)
```

# **ABOVE-MEAN SQUARES PROBLEM**

```python
S = (np.arange(1, 37) ** 2).reshape(6, 6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
above_mean.size
```

```python
np.save('above_mean.npy', above_mean)
```
